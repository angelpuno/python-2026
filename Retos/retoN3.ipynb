```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python

def preparar_datos():
    horas_simulacion= np.arange(1,11)
    CPU_registrado= pd.array([45, 58, 52, 70, 85, 92, 78, 64, 61, 50])

    data={
        "h_simulacion": horas_simulacion,
        "CPU_registrado": CPU_registrado
    }

    df= pd.DataFrame(data)
    return df
```

```python
def graficar_rendimiento(df):
    fig, ax = plt.subplots()
    ax.plot(df["h_simulacion"], df["CPU_registrado"], marker="o", color="red", linestyle="--")
    ax.set_title("Monitoreo de la Infraestructura Tecnologica", fontsize= 20)
    ax.set_xlabel("Horas de Simulacion")
    ax.set_ylabel("Uso de CPU_Registrado")
    ax.grid(True)
    fig.savefig("grafico01", dpi=300)
```

```python
df_CPU_simulacion= preparar_datos()
graficar_rendimiento(df_CPU_simulacion)
plt.show()
```
