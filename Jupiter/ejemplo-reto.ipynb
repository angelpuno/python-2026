Como ingenieros/as informáticos/as, se les ha encomendado procesar los datos de medición de caudal (en metros cúbicos por segundo, m³/s) tomados de forma automatizada cada 1 hora durante un turno de 8 horas.

Construir y devolver un DataFrame de Pandas que contenga dos columnas denominadas:
"Hora" y "Caudal (m3/s)"

2. Función visualizar_curva(dataframe)

Debe recibir como parámetro el DataFrame generado por la función anterior.

Utilizando Matplotlib, generar un gráfico de líneas para observar la curva de crecida del río.

el grafico debe incorporar:
marker= "o"

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def procesar_caudal():
    horas= np.arange(1,9)
    caudales=[12.5 , 14.2 , 18.1 , 22.8 , 25.4 , 21.0, 17.5, 15.3]

    data={
        "hora": horas,
        "caudal": caudales
    }

    df= pd.DataFrame(data)
    return df
```

```python
def viualizar_curva(dataframe):
    fig, ax = plt.subplots()
    ax.plot(dataframe["hora"], dataframe["caudal"], marker="o", color="green", linestyle="-")
    ax.set_title("Monitoreo de Caudal Rio Gamboa", fontsize= 20)
    ax.set_xlabel("Tiempo (Horas)")
    ax.set_ylabel("Caudal (m^3/s)")
    ax.grid(True)
```

```python
df_hidrologia= procesar_caudal()
viualizar_curva(df_hidrologia)
plt.show()
```
